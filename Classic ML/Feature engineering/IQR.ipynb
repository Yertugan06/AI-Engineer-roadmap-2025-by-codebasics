```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('weight-height.csv')
```

```python
df.info()
```

```python
df.describe()
```

```python
fig, ax = plt.subplots(2, 1, figsize=(10, 5))
sns.histplot(df['Height'], color='red', ax = ax[0], kde=True, bins = 20)
sns.histplot(df['Weight'], color='blue', ax = ax[1], kde = True, bins = 20)
plt.show()
```

```python
Q1 = df.Height.quantile(0.25)
Q3 = df.Height.quantile(0.75)
IQR = Q3 - Q1
upper_limit = Q3 + 1.5*(IQR)
lower_limit = Q1 - 1.5*(IQR)
df2 = df[(df.Height < upper_limit) & (df.Height > lower_limit)]
```

```python
Q1 = df.Weight.quantile(0.25)
Q3 = df.Weight.quantile(0.75)
IQR = Q3 - Q1
upper_limit = Q3 + 1.5*(IQR)
lower_limit = Q1 - 1.5*(IQR)
df3 = df[(df.Weight < upper_limit) & (df.Weight > lower_limit)]
```

```python
fig, ax = plt.subplots(2, 1, figsize=(10, 5))
sns.histplot(df2['Height'], color='red', ax = ax[0], kde=True, bins = 20)
sns.histplot(df3['Weight'], color='blue', ax = ax[1], kde = True, bins = 20)
plt.show()
```
