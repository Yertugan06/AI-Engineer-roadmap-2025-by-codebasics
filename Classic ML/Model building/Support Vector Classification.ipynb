```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.datasets import load_digits
```

```python
digits = load_digits()
```

```python
dir(digits)
```

```python
df = pd.DataFrame(digits.data,columns = digits["feature_names"])
df.head()
```

```python
df["target"] = digits.target
df.head()
```

```python
df.info()
```

```python
x_train, x_test, y_train, y_test = train_test_split(df.drop("target",axis = "columns"),df.target,test_size = 0.2, random_state = 42)
```

```python
model = SVC()
```

```python
param_grid_svm = {
    'C': [0.1, 1, 10, 100],
    'gamma': [0.001, 0.01, 0.1, 1],
    'kernel': ['linear', 'rbf', 'poly', 'sigmoid'],
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
grid_search = GridSearchCV(estimator = model, param_grid=param_grid_svm)
```

```python
grid_search.fit(x_train,y_train)
```

```python
model.fit(x_train,y_train)
```

```python
best_model = grid_search.best_estimator_
print("Test score with best parameters:", best_model.score(x_test, y_test))
print("Sk learn result", model.score(x_test,y_test))
```
