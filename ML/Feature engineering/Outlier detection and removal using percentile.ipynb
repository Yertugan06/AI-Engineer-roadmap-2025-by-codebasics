```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('AB_NYC_2019.csv')
df.sort_values('price', ascending=False, inplace = True)
df.head(10)
```

```python
df.sort_values('price', ascending=True, inplace = True)
df.head(10)
```

```python
df.price.describe()
```

```python
max_threshold = df['price'].quantile(0.999)
min_threshold = df['price'].quantile(0.01)
print(max_threshold)
print(min_threshold)
```

```python
df2 = df[(df['price'] < max_threshold) & (df['price'] > min_threshold)]
df2.sample(5)
```

```python
df2.price.describe()
```
